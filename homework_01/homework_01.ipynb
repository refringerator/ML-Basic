## Задание 1: Зеркальное отражение цифр

Пользователь вводит пятизначное число. Программа должна зеркально отразить три центральные цифры. Первая и последняя остаются на местах.

<br>

**Примеры:**
* `23456 -> 25436`
* `30789 -> 38709`

```python
num = input("Введите число для преобразования: ").strip()
if len(num) < 3:
    print("Число должно быть не менее 3-х знаков")
else:
    result = num[0] + num[-2:0:-1] + num[-1]
    print(f"Для '{num}' переделанное число '{result}'")
```

## Задание 2: Подсчет выходных до отпуска

Пользователь вводит сколько дней осталось до ближайшего отпуска. Программа должна вывести количество выходных дней до отпуска, если учесть, что выходные это суббота и воскресенье, сегодня понедельник и праздники мы не учитываем.  

<br>

**Примеры:**
* `4  -> 0`
* `6  -> 1`
* `14 -> 4`

```python
num = input("Введите сколько дней до отпуска: ")
try:
    num = int(num)
    if num < 0:
        raise ValueError("Отрицательное число")
    
    weeks, days = divmod(num, 7)
    result = weeks * 2 + max(days - 5, 0)
    print(f"Если до отпуска {num} дн., то количество полных выходных до отпуска: {result}")

except ValueError:
    print("Введите целое положительное число")
    
```

## Задание 3: Плитка шоколада

Пользователь вводит длину и ширину плитки шоколада, а также размер куска, который хочет отломить. Программа должна вычислить - можно ли совершить подобный разлом или нет, если учесть, что ломать можно только по прямой.

<br>

**Примеры:**
* `3, 4, 6  -> True`
* `5, 7, 8  -> False`
* `4, 5, 12 -> True`

```python
try:
    user_input = input("Введите длину, ширину и размер куска шоколадки (3 числа через запятую): ").split(",")
    length, width, size = [int(n) for n in user_input if int(n) > 0]
    choco_enough = size <= width * length
    can_break = size % width == 0 or size % length == 0
    print("Можно" if choco_enough and can_break else "Нельзя")

except ValueError:
    print("Введите три положительных числа через запятую")
```

## Задание 4: Римские числа

Пользователь вводит целое положительное число, программа должна вернуть строку в виде римского числа.

<br>

**Примеры:**
* `3   -> III`
* `15  -> XV`
* `234 -> CCXXXIV`

```python
r = {
    'I': 1,
    'IV': 4,
    'V': 5,
    'IX': 9,
    'X': 10,
    'XL': 40,
    'L': 50,
    'XC': 90,
    'C': 100,
    'CD': 400,
    'D': 500,
    'CM': 900,
    'M': 1000,
}

try:
    num = int(input("Введите натуральное число: "))
    if num < 1:
        raise ValueError("Число меньше 1го")
    if num % 1 != 0:
        raise ValueError("Не целое число")
    if num > 3999:
        raise ValueError("Число больше 3999")
    
    result = []
    rem = num
    for roman, value in reversed(r.items()):
        count, rem = divmod(rem, value)
        result.append(roman * count)
    print(f"Для числа {num} римское представление: {''.join(result)}")

except ValueError:
    print("Введите целое положительное число")
```

## Задание 5: Проверка на вещественное число

Пользователь вводит данные, проверить - являются ли они положительным вещественным числом. Не использовать встроенные функции для проверки, только методы данных и конструкцию IF. (Дополнительное задание, по желанию - проверка на отрицательные вещественные числа)

<br>

**Примеры:**
* `5.6  -> True`
* `.78  -> True`
* `.67. -> False`
* `5    -> True`

```python
num = input("Введите вещественное число: ").strip()
try:
    if negative := num.startswith('-'):
        num = num[1:]
    
    if len(num) < 1:
        raise ValueError("Пустая строка")
    if num.count('.') > 1 or num.endswith('.'):
        raise ValueError("Больше одной точки или точка в конце")
    
    for digit in num:
        if digit not in '0123456789.':
            raise ValueError("Есть нецифровые символы")

    print(f"Введено {'отрицательное ' if negative else ''}вещественное число")

except ValueError:
    print("Введено не вещественное число")
```
