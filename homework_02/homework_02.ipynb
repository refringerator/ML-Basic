## Задание 1: Получение однозначного числа

Пользователь вводит целое число, программа складывает все цифры числа, с полученным числом — то же самое, и так до тех пор, пока не получится однозначное число.

<br>

**Примеры:**
* `545   -> 5`
* `12345 -> 6`

````python
def compact(num: int) -> int:
    result = sum([int(digit) for digit in str(num)])
    return result if result < 10 else compact(result)

try:
    num = int(input("Введите целое число: "))
    result = compact(abs(num))
    print(f"Компактная форма числа {num} это {result}") 

except ValueError:
    print("Нужно ввести целое число")
````

## Задание 2: Кинотеатр

Дан список списков, каждый вложенный список состоит из 1 и 0, количество вложенных списков — количество рядов. Пользователь вводит, сколько билетов ему требуется. Программа должна найти ряд, где можно приобрести нужное количество билетов (места должны быть рядом). Если таких рядов несколько, то ближайший к экрану (ближайшим считается нулевой ряд). Если таких мест нет, то вывести False.

<br>

**Примеры:**
* `[[0,1,1,0], [1,0,0,0], [0,1,0,0]], 2 -> 1`
* `[[0,1,1,0], [1,0,1,0], [1,1,0,1]], 2 -> False`

````python
from typing import Literal

examples = [
    ([[0,1,1,0], [1,0,0,0], [0,1,0,0]], 2, 1),
    ([[0,1,1,0], [1,0,1,0], [1,1,0,1]], 2, False),
]

def check_row(row: list, count: int) -> bool:
    return count * "0" in "".join(map(str, row))

def check_hall(hall: list[list[int]], count: int) -> int | Literal[False]:
    for row_index, row in enumerate(hall):
        if check_row(row, count):
            return row_index
    return False

for hall, count, expected in examples:
    result = check_hall(hall, count)
    print(f"Для {hall} и {count} ожидалось {expected}, получено {result}")
    
    assert result == expected
````

## Задание 3: Алгоритм RLE

Необходимо написать упрощенную версию алгоритма RLE. 
Алгоритм RLE объединяет подряд идущие символы в коэффициент и символ.

<br>

**Примеры:**
* `asssdddsssddd -> 1a3s3d3s3d`
* `aaabbbbccccc  -> 3a4b5c`
* `abcba         -> 1a1b1c1b1a`

````python
from itertools import groupby

examples = [
    ("asssdddsssddd", "1a3s3d3s3d"),
    ("aaabbbbccccc",  "3a4b5c"),
    ("abcba",         "1a1b1c1b1a"),
    ("",              ""),
]


def simple_rle(data: str) -> str:
    result = [f"{len(list(group))}{key}" for key, group in groupby(data)]
    return "".join(result)

    
for data, expected in examples:
    result = simple_rle(data)
    print(f"Для '{data}' ожидалось '{expected}', получено '{result}'")
    
    assert result == expected

````

## Задание 4: Шифр Цезаря

Реализуйте программу для шифрования текста с помощью шифра Цезаря.  
Шифр Цезаря — это метод шифрования, при котором каждая буква в тексте заменяется буквой, стоящей на фиксированное число позиций дальше в алфавите.

### Требования
1. Программа должна принимать на вход:
   - Строку текста для шифрования
   - Ключ шифра (число от 0 до 25)

2. Правила шифрования:
   - Шифруются только латинские буквы (a-z, A-Z)
   - Сохраняется регистр букв (заглавные остаются заглавными)
   - Пробелы, цифры и знаки препинания остаются без изменений
   - Сдвиг циклический (после 'z' идет 'a', после 'Z' идет 'A')

### Примеры

* `"Hello World", 3      -> "Khoor Zruog"`
* `"Python is great!", 5 -> "Udymts nx lwjfy!"`
* `"XYZ abc", 3          -> "ABC def"`
* `"Test 123", 1         -> "Uftu 123"`

### Алгоритм
* Для каждого символа в строке:
   - Если это латинская буква, найти её позицию в алфавите
   - Добавить к позиции значение ключа
   - Если результат больше 25, использовать остаток от деления на 26
   - Преобразовать обратно в букву, сохранив регистр
   - Если это не буква, оставить символ без изменений


````python
examples = [
    ("Hello World",      3, "Khoor Zruog"),
    ("Python is great!", 5, "Udymts nx lwjfy!"),
    ("XYZ abc",          3, "ABC def"),
    ("Test 123",         1, "Uftu 123"),
    ("bcd 123",         -1, "abc 123"),
]

from string import ascii_lowercase, ascii_uppercase


class ShiftedDict(dict):
    def __init__(self, shift: int):
        super().__init__()
        
        shift %= len(ascii_lowercase)
        shifted = ascii_lowercase[shift:] + ascii_lowercase[:shift]

        self.update(dict(zip(ascii_lowercase, shifted)))
        self.update(dict(zip(ascii_uppercase, shifted.upper())))


def chiper(text: str, shift: int) -> str:
    table = str.maketrans(dict(ShiftedDict(shift)))
    return text.translate(table)


for data, shift, expected in examples:
    result = chiper(data, shift)
    print(f"Для '{data}' со сдвигом {shift} ожидалось '{expected}', получено '{result}'")
    
    assert result == expected
````

## Задание 5: Табель успеваемости

Пользователь в бесконечном цикле (пока не будет введена пустая строка) вводит строки вида: 'название предмета' 'фамилия ученика' 'оценка'. После окончания ввода программа выводит в консоль название предмета, далее список учеников и все их оценки в виде таблицы.

<br>

**Примеры:**

<br>

*Ввод:*
```python
Математика Иванов  5
Математика Иванов  4
Литература Иванов  3
Математика Петров  5
Литература Сидоров 3
Литература Петров  5
Литература Иванов  4
Математика Сидоров 3
Математика Петров  5
```

<br>

*Результат:*
```python
Математика # вводим название предмета

# выводим список учеников и их оценки
Иванов  5 4
Петров  5 5
Сидоров 3
```

```python
Литература # вводим название предмета

# выводим список учеников и их оценки
Иванов  3 4
Сидоров 3
Петров  5
```

````python
from collections import defaultdict


journal = defaultdict(lambda: defaultdict(list))


def ascii_table(journal):

    table_width = 0
    student_width = 0
    score_width = 0
    
    def width(items):
        return len(max(items, key=len))
    
    def line(char="-", sep="+"):
        return sep + char * table_width + sep
    
    def row(student, scores):
        return f"| {student:<{student_width}} | {scores:<{score_width}} |"
    
    def header(cls):
        return f"{line('=')}\n| {cls:<{table_width - 2}} |\n{line('=')}"
    
    
    def calulate_widths():
        nonlocal table_width, student_width, score_width
        
        cls_width = width(journal.keys())
        student_width = width(
            (name for students in journal.values() for name in students.keys())
        )
        score_width = width(
            (" ".join(scores) for students in journal.values() for scores in students.values())
        )
        
        table_width = max(cls_width, student_width + score_width + 5)
    
    def print_table():
        for cls, students in sorted(journal.items()):
            print(header(cls))
            for name, scores in sorted(students.items()):
                print(row(name, " ".join(scores)))
        print(line())

    calulate_widths()
    print_table()



while(True):
    try:
        user_input = input("Введите строку: ")
        if not user_input:
            break
        cls, name, score = user_input.split()

        journal[cls][name].append(score)

    except ValueError:
        pass

ascii_table(journal)
````
