## Задание 1: Конвертер регистров

Написать функцию, которая будет переводить snake_case в PascalCase и наоборот. 

Функция должна сама определять - какой формат ей передали. Можно добавить ключевой аргумент, который будет принудительно возвращать один из форматов.

<br>

**Примеры:**
* `otus_course     -> OtusCourse`
* `PythonIsTheBest -> python_is_the_best`



```python
from typing import Literal


def convert(text: str, format: Literal["auto", "snake", "pascal"] = "auto") -> str:
    if format == "auto":
        if "_" in text:
            format = "pascal"
        else:
            format = "snake"

    if format == "pascal":
        parts = text.split("_")
        return "".join(word.capitalize() for word in parts)

    elif format == "snake":
        result = []
        word_start = 0
        for i, char in enumerate(text):
            if char.isupper() and i != 0:
                result.append(text[word_start:i].lower())
                word_start = i
        result.append(text[word_start:].lower())
        return "_".join(result)


print(convert("otus_course", "auto"))
print(convert("PythonIsTheBest", "auto"))
```

## Задание 2: Проверка валидности даты

Написать функцию проверяющую валидность введенной даты.

<br>

**Примеры:**
* `29.02.2000 -> True`
* `29.02.2001 -> False`
* `31.04.1962 -> False`



```python
examples = [
    ("29.02.2000", True),
    ("29.02.2001", False),
    ("31.04.1962", False),
]


def is_leap_year(year: int) -> bool:
    return (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)


def check_date(date_str: str) -> bool:
    day, month, year = map(int, date_str.split("."))

    if month < 1 or month > 12:
        return False

    month_days = [31, 28 + is_leap_year(year), 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]

    if day < 1 or day > month_days[month - 1]:
        return False

    return True


for date_str, expected in examples:
    result = check_date(date_str)
    print(
        f"Проверяемая дата: {date_str}, результат : {result}, Ожидаемый результат: {expected}"
    )

    assert result == expected
```

## Задание 3: Проверка на простое число

Функция проверки на простое число. Простые числа – это такие числа, которые делятся на себя и на единицу.

<br>

**Примеры:**
* `17 -> True`
* `20 -> False`
* `23 -> True`

```python
examples = [
    ("17", True),
    ("20", False),
    ("23", True),
]


def is_prime(num: int) -> bool:
    if num < 2:
        return False

    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False

    return True


for num_str, expected in examples:
    result = is_prime(int(num_str))
    print(
        f"Проверяемое число: {num_str}, результат : {result}, Ожидаемый результат: {expected}"
    )

    assert result == expected
```

## Задание 4: Учет пользователей

Пользователь в бесконечном цикле вводит данные пользователей: имя, затем фамилию, возраст и ID. Ввод продолжается до тех пор, пока не будет введено пустое поле. 

Пользователи заносятся в словарь, где ключ это ID пользователя, а остальные данные записываются в виде кортежа. 

**Программа должна проверять:**
* имя и фамилия состоят только из символов и начинаются с большой буквы - если не с большой, то заменяет букву на большую;
* возраст должен быть числом от 18 до 60;
* ID - целое число, дополненное до 8 знаков незначащими нулями, ID должен быть уникальным.

**Дополнительно:** написать функцию, которая будет выводить полученный словарь в виде таблицы.

```python
input_data = [
    "Мария Тихонова 6 1",
    "Михаил Лебедев 1 2",
    "Кирилл Панфилов 3 3",
    "Глеб Карпов 6 4",
    "Станислав Никуличев 4 5",
    "Николай Осипов 2 6",
    "Константин Алексин 4 7",
    "Евгений Ревняков 7 8",
    "Евгений Романов 8 9",
    "Игорь Стурейко 3 10",
    "Без Опыта 0 11",
    "Дублирующий Идентификатор 1 1",
    "",
]


from dataclasses import dataclass, fields


@dataclass(frozen=True, slots=True)
class Record:
    name: str
    surname: str
    age: int

    def __post_init__(self):
        if self.age < 1 or self.age > 60:
            raise ValueError("Опыт должен быть в диапазоне от 1 до 60 лет.")

        if not isinstance(self.name, str) or not isinstance(self.surname, str):
            raise TypeError("Имя и фамилия должны быть строками.")

        if not self.name or not self.surname:
            raise ValueError("Имя и фамилия не могут быть пустыми.")


class Database:
    def __init__(self, id_length: int = 8):
        self.id_length = id_length
        self.data: dict[str, Record] = {}

    def __iter__(self):
        return iter(self.data.items())

    def convert_id(self, identifier: int) -> str:
        return str(identifier).zfill(self.id_length)

    def is_id_exist(self, identifier: int) -> bool:
        return self.convert_id(identifier) in self.data

    def add_record(self, identifier: int, record: Record):
        self.data[self.convert_id(identifier)] = record

    def print_ascii_table(self) -> None:
        row_template = ""
        table_width = 0
        cols_width = {}

        def width(items):
            return len(max(items, key=len))

        def line(char="-", sep="+"):
            return sep + char * table_width + sep

        def row(identifier: str, record: Record):
            return row_template.format(identifier=identifier, record=record)

        def header():
            return (
                f"{line('=')}\n| "
                + f"{'ID':^{self.id_length}} | "
                + " | ".join(
                    [
                        f"{field.name.title():^{cols_width[field.name]}}"
                        for field in fields(Record)
                    ]
                )
                + f" |\n{line('=')}"
            )

        def calulate_widths():
            nonlocal row_template, table_width, cols_width

            for field in fields(Record):
                cols_width[field.name] = max(
                    len(field.name),
                    width(
                        str(getattr(record, field.name))
                        for record in self.data.values()
                    ),
                )

            table_width = (
                self.id_length + 3 + sum(cols_width.values()) + 3 * len(cols_width) - 1
            )

            row_template = (
                f"| {{identifier:<{self.id_length}}} | "
                + " | ".join(
                    [
                        f"{{record.{field.name}:{'>' if field.type == int else '<'}{cols_width[field.name]}}}"
                        for field in fields(Record)
                    ]
                )
                + " |"
            )

        def print_table():
            print(header())
            for identifier, record in sorted(self.data.items()):
                print(row(identifier, record))
            print(line())

        calulate_widths()
        print_table()


db = Database()

while True:
    try:
        # user_input = input("Введите строку: ")
        user_input = input_data.pop(0)
        if not user_input:
            break
        name, surname, age, i = user_input.split()

        record = Record(name, surname, int(age))
        identifier = int(i)
        if db.is_id_exist(identifier):
            print(f"Запись с идентификатором {identifier} уже существует.")
        else:
            db.add_record(identifier, record)

    except ValueError:
        pass

db.print_ascii_table()
```
